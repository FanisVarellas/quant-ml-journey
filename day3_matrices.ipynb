```python
import numpy as np

```

```python
A=np.array([[2,0],[0,2]])
V=np.array([3,1])
print(A@V)
```

```python
B = np.array([[1, 2], [0, 3]])
w = np.array([1, 1])
print(B @ w)
```



What a matrix is (in terms of i and j): a matrix records where i and j moved to, that record (the two columns) is the transformation.

What A @ v actually calculates:  the new point the transformation moved the vector.

The row-versus-column trap you just hit, and how you now know to avoid confusion (hint: you do not need to hand-flip it, @ does it correctly, but you understand what is inside): you think in terms of "x amount of i, y amount of j", and you now know NumPy's [[...],[...]] stores rows, so the actual new-i and new-j are the columns of that grid, not the inner lists themselves. You do not need to hand-flip it because @ handles it, but you know what is happening underneath. That is real understanding, not surface-level pattern matching.
